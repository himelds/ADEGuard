```python
import json

with open('../data/processed/exports/annotation_50.json', 'r') as f:
    data = json.load(f)

drug_counts = []
ade_counts = []

for item in data:
    annotations = item['annotations'][0]['result']
    drugs = [a for a in annotations if a['value']['labels'][0] == 'DRUG']
    ades  = [a for a in annotations if a['value']['labels'][0] == 'ADE']
    drug_counts.append(len(drugs))
    ade_counts.append(len(ades))

print(f"Avarage DRUG per report: {sum(drug_counts)/len(drug_counts):.2f}")
print(f"Avarage ADE per report: {sum(ade_counts)/len(ade_counts):.2f}")
print(f"nonADE Report: {ade_counts.count(0)}")
print(f"nonDRUG Rport: {drug_counts.count(0)}")
```
