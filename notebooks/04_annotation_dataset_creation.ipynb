```python
import pandas as pd
```

```python
df = pd.read_csv('../data/processed/vaers_clean_dataset.csv')
```

```python
df.head()
```

```python
sample = pd.concat([
    df[df['SEVERITY_LEVEL'] == 'Severe'].sample(200, random_state=42),
    df[df['SEVERITY_LEVEL'] == 'Moderate'].sample(300, random_state=42),
    df[df['SEVERITY_LEVEL'] == 'Mild'].sample(300, random_state=42),
])

sample = sample.sample(frac=1, random_state=42).reset_index(drop=True)

sample.to_csv("../data/processed/annotation_batch_final.csv", index=False)

print(sample['SEVERITY_LEVEL'].value_counts())
print(f"Total: {len(sample)}")
```
